- `(N,D) @ (D,K) = (N,K) ~ Basically`
    - N = sample 
    - D = in_dim
    - K = out_dim 
    - in linear regresison K = 1 

X ~ (N, D+1)
W ~ (D+1, )
X@W is (N,)


A network is just that rule chained, where each layer's output width becomes the next layer's input width:

(N, 784) @ (784, 128) → (N, 128) @ (128, 64) → (N, 64) @ (64, 10) → (N, 10)

N never changes. It rides along the whole way. Only the feature dimension gets rewritten at each step, and the last one has to match whatever y looks like. That's the entire shape story of a feedforward network.
